```python

# Keep only rows whose raceId-driverId combination repeats
repeated_pairs = df[
    df.duplicated(
        subset=["raceId", "driverId"],
        keep=False
    )
].copy()

# Count unique race-driver combinations
number_of_pairs = repeated_pairs[
    ["raceId", "driverId"]
].drop_duplicates().shape[0]

print("Rows involved in repeated pairs:", len(repeated_pairs))
print("Number of unique repeated race-driver pairs:", number_of_pairs)

# Count how many times each pair appears
pair_counts = (
    repeated_pairs
    .groupby(["raceId", "driverId"])
    .size()
    .reset_index(name="number_of_rows")
    .sort_values("number_of_rows", ascending=False)
)

display(pair_counts.head(20))
```

```python
df_clean = df.copy()
df_clean = df_clean.replace(r"\N",pd.NA)
print(df_clean.isna().sum().sort_values(ascending=False))
```

```python
print(df_clean.dtypes)
```

```python
numeric_columns = [
    "number",
    "grid",
    "position",
    "milliseconds",
    "fastestLap",
    "rank",
    "fastestLapSpeed"
]
for column in numeric_columns :
    df_clean[column] = pd.to_numeric(df_clean[column],errors= "coerce")
```

```python
df_clean[numeric_columns].dtypes
```

```python
df_clean[numeric_columns].isna().sum().sort_values(ascending=False)
```

```python
for column in numeric_columns:
    negative_count = (df_clean[column] < 0).sum()
    print(f"{column}: {negative_count} negative values")
```

```python
df_clean.describe().T
```

```python
position_comparaison = df_clean[df_clean["position"].notna()].copy()    #dataframe
different_positions = position_comparaison["positionOrder"] != position_comparaison["position"]   #boolean series
print ("Rows where position is available:", len(position_comparaison))
print("Rows where position differs from positionOrder:",different_positions.sum())
display(position_comparaison.loc[different_positions,["raceId", "driverId", "position","positionText", "positionOrder", "statusId"]].head(15))
```

```python

# Inspect the distinct values and their frequencies
categorical_columns = [
    "positionText",
    "time",
    "fastestLapTime"
]

for column in categorical_columns:
    print(f"\n--- {column} ---")
    print("Number of distinct values:", df_clean[column].nunique())
    print("\nMost frequent values:")
    print(df_clean[column].value_counts(dropna=False).head(15))
```

```python
missing_count = df_clean.isna().sum()

missing_percentage = (
    df_clean.isna().mean() * 100
).round(2)

missing_summary = pd.DataFrame({
    "Missing_Count": missing_count,
    "Missing_Percentage": missing_percentage
})

missing_summary = missing_summary.sort_values(
    by="Missing_Percentage",
    ascending=False
)

missing_summary
```
