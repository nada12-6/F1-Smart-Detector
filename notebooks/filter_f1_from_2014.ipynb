```python

import pandas as pd
df = pd.read_csv("../data/processed/f1_merged.csv")
grid_by_race = {
    # Qatar 2025 : Colapinto part de la pit lane, donc grid = 0
    1167: {
        "PIA": 1, "NOR": 2, "VER": 3, "RUS": 4, "ANT": 5,
        "HAD": 6, "SAI": 7, "ALO": 8, "GAS": 9, "LEC": 10,
        "HUL": 11, "LAW": 12, "BEA": 13, "ALB": 14, "TSU": 15,
        "OCO": 16, "HAM": 17, "STR": 18, "BOR": 19, "COL": 0,
    },

    # Azerbaïdjan 2026 : grille finale après pénalités
    1183: {
        "RUS": 1, "LEC": 2, "PIA": 3, "HAD": 4, "NOR": 5,
        "HAM": 6, "GAS": 7, "VER": 8, "COL": 9, "BEA": 10,
        "LAW": 11, "ALB": 12, "OCO": 13, "SAI": 14, "LIN": 15,
        "ANT": 16, "BOR": 17, "HUL": 18, "BOT": 19, "PER": 20,
        "ALO": 21, "STR": 22,
    },
}

for race_id , position in grid_by_race.items():
    race_rows= df["race_id"].eq(race_id)
    missing_grid = race_rows & df["grid"].isna()
    df.loc[missing_grid, "grid"] = (df.loc[missing_grid,"driver_code"].map(position))
    if df.loc[missing_grid,"grid"].isna().any():
        raise ValueError(f"Grid FIA manquant pour race {race_id}")

df["grid"] = pd.to_numeric(df["grid"], errors="coerce").astype("Int64")


df = df[df["year"] >= 2014].copy()

print(df.shape)
print(df["year"].min())
print("Grid manquants :", df["grid"].isna().sum())
print("Grid = 0 :", (df["grid"] == 0).sum())
print("Grid max :", df["grid"].max())
df.to_csv("../data/processed/f1_merged_2014.csv", index=False)

```
